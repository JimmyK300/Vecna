```python
# import these pls ;3
from transformers import pipeline 
from PIL import Image

sigCLIP = pipeline(task="zero-shot-image-classification", model="google/siglip2-base-patch16-224", device=0)
```

```python
car_image = Image.open("..\\data\\car_image.jpg")
cat_image = Image.open("..\\data\\cat_image.jpg")
dog_image = Image.open("..\\data\\dog_image.jpg")

images = [car_image, cat_image, dog_image]
descs = ["car", "cat", "dog"]

# labels = ["a photo of a cat", "a photo of a dog", "a photo of a car"]
query = ["a picture of a car"] # try animal, fluffy object, mammals, golden retriver, mèo
for image, desc in zip(images, descs):
    print(desc, end=" ")
    result = sigCLIP(image, candidate_labels=query)
    print(result)
```
